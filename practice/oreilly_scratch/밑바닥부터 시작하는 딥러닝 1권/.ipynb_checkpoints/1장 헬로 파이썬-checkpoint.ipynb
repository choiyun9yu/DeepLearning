### 1-1. Class

```python
class Man:
    def __init__(self, name):
        self.name = name
        print("Initialized!")

    def hello(self):
        print("Hello " + self.name + "!")

    def goodbye(self):
        print("Good-bye " + self.name + "!")
```

```python
m = Man("David")
m.hello()
m.goodbye()
```

## 1-2. Numpy
- 딥러닝을 구현하다 보면 배열이나 행렬 계산이 많이 등장한다.
- 넘파이의 배열 클래스인 numpy.array 에는 편리한 메서드가 많이 준비되어 있다.
- 딥러닝을 구현할 때 이 메서드들을 이용한다.

```python
# 넘파이 가져오기 
import numpy as np

# 넘파이 배열 생성하기
x = np.array([1.0, 2.0, 3.0])
print(x)

type(x)
```

```python
# 넘파이 산술 연산 
x = np.array([1.0, 2.0, 3.0])
y = np.array([2.0, 4.0, 6.0])

print("원소별 덧셈: ", x + y)

print("원소별 뺄셈: ", x - y)

print("원소별 곱셈: ", x * y)

print("원소별 나눗셈: ", x / y)
```

- 여기서 주의할 점은 배열 x 와 y 의 원소 수가 같다는 것이다.
    - 원소 수가 같으면, 산술 연산은 각 원소에 대해 행해진다.
    - 원소 수가 다르면, 오류가 발생하니 원소 수 맞추기는 매우 중요하다.
    - 참고로 "원소별" 이라는 말은 영어로 element-wise 이다.  
      원소별 곱셈은 element-wise product 라고 한다.

- 넘파이 배열은 원소별 계산 뿐 아니라 넘펴아 배열과 수치 하나(스칼라값)의 조합으로 된 산술연산도 수행할 수 있다.
- 이 경우 스칼라갑소가의 계싼이 넘파이 배열의 원소별로 한 번 씩 수행된다. 그리고 이 기능을 브로드캐스트라고 한다.

```python
# 브로드캐스트
x = np.array([1.0, 2.0, 3.0])
x / 2.0
```

```python
# 넘파이의 N차원 배열
A = np.array([[1, 2], 
              [3, 4]])

print(A)

print("shape: ", A.shape)

print("dtype: ", A.dtype)
```

#### 행렬 곱셈 


```python
# 넘파이 N차원 배열의 산술연산
A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[3, 0], 
              [0, 6]])

print("A + B: \n", A + B)

print("----------")

print("A * B: \n", A * B)

print("----------")

print("A * 10: \n", A * 10)
```

### 브로드캐스트
- 넘파이에서는 형상이 다른 배열끼리도 계산을 할 수 있다.
- 앞의 예에서는 2x2 행렬 A 에 스칼라 값 10을 곱했다. 이 경우 10이라는 스칼라값이 2x2 행렬로 확대된다.
  


```python
A = np.array([[1, 2], 
              [3, 4]])

B = np.array([10, 20])

A * B
```

```python
# 원소 접근
X = np.array([[51, 55],
              [14, 19],
              [0, 4]])
print(X)

print("----------")

print(X[0])

print("----------")

print(X[0][1])
```

```python
for row in X:
    print(row)
```

```python
# X 를 1 차원 배열로 평탄화
X = X.flatten()

print(X)
```

- 이 기법을 응용하면 특정 조건을 만족하는 원소만 넣을 수 있다.
    - 예를 들어 X 에서 15 이상인 값만 구할 수 있다.
    - 넘파이 배열에 부동호 연산자를 사용한 결과는 bool 배열이다.
    - 여기에 bool 배열을 사용해서 X 에서 True 에 해당하는 원소, 즉 15보다 큰 원소만 꺼낼 수 있다.

```python
X > 15
```

```python
X[X>15]
```

## 1-3. Matplotlib

### pyplot moudule
- 그래프를 그릴 때 사용한다.

```python
import matplotlib.pyplot as plt

# 데이터 준비
x = np.arange(0, 6, 0.1)  # 0에서 6까지 0.1 간격으로 생성
y = np.sin(x)  # sin 함수

# 그래프 그리기
plt.plot(x, y)
plt.show()
```

```python
# 데이터 준비
x = np.arange(0, 6, 0.1)
y1 = np.sin(x)
y2 = np.cos(x)

# 그래프 그리기
plt.plot(x, y1, label="sin")  # sin 함수 곡선 그리기
plt.plot(x, y2, linestyle="--", label="cos")  # cos 함수 점선 그리기
plt.xlabel("x")  # x 축 이름 
plt.ylabel("y")  # y 축 이름
plt.title('sin & cos')  # 제목
plt.legend()  # 범례
plt.show()
```
